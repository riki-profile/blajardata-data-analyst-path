## 3. Tipe data

| Tipe | Nama di Python | Contoh |
| --- | --- | --- |
| Bilangan bulat | `int` | `125`, `-3` |
| Bilangan desimal | `float` | `49.99`, `0.15` |
| Teks | `str` | `"Jeans"`, `'Complete'` |
| Benar/salah | `bool` | `True`, `False` |

Gunakan `type()` untuk mengecek tipe sebuah nilai.

````python
status = "Complete"
sudah_dikirim = True

print(type(jumlah_pesanan))
print(type(harga_jeans))
print(type(status))
print(type(sudah_dikirim))
````

### Operasi angka

| Operator | Arti | Contoh | Hasil |
| --- | --- | --- | --- |
| `+ - * /` | tambah, kurang, kali, bagi | `10 / 4` | `2.5` |
| `//` | bagi, dibulatkan ke bawah | `10 // 4` | `2` |
| `%` | sisa bagi | `10 % 4` | `2` |
| `**` | pangkat | `2 ** 3` | `8` |

````python
harga = 49.99
jumlah_barang = 3
diskon_persen = 10

subtotal = harga * jumlah_barang
diskon = subtotal * diskon_persen / 100
total = subtotal - diskon

print(subtotal)
print(diskon)
print(round(total, 2))   # round(angka, jumlah_desimal)
````

### Teks dan f-string

Cara paling praktis menggabungkan teks dan angka adalah **f-string**: tulis `f` sebelum tanda kutip, lalu taruh variabel di dalam `{ }`.

````python
kategori = "Jeans"
pendapatan = 1234567.891

print(f"Pendapatan kategori {kategori}: {pendapatan} USD")
print(f"Pendapatan kategori {kategori}: {pendapatan:,.2f} USD")   # :,.2f = pemisah ribuan, 2 desimal
````

````python
# Beberapa operasi teks yang berguna
brand = "  calvin klein  "
print(brand.strip())           # buang spasi di awal & akhir
print(brand.strip().title())   # Huruf Awal Kapital
print(brand.upper())           # HURUF BESAR
print(len("TheLook"))          # jumlah karakter
````

## 4. List: kumpulan nilai berurutan

List ditulis dengan kurung siku `[ ]`. Misalnya, harga jual 6 barang dalam pesanan hari ini:

````python
harga_terjual = [49.99, 15.00, 120.50, 32.75, 8.99, 75.00]
print(harga_terjual)
````

### Mengambil isi list dengan indeks

Indeks **dimulai dari 0**. Indeks negatif dihitung dari belakang.

```
harga_terjual = [49.99, 15.00, 120.50, 32.75, 8.99, 75.00]
indeks:            0      1      2       3      4     5
                  -6     -5     -4      -3     -2    -1
```

````python
print(harga_terjual[0])    # barang pertama
print(harga_terjual[2])    # barang ketiga
print(harga_terjual[-1])   # barang terakhir
print(harga_terjual[0:3])  # slicing: indeks 0, 1, 2 (angka akhir tidak ikut)
````

### Fungsi bawaan untuk list

````python
print("Jumlah barang :", len(harga_terjual))
print("Total         :", sum(harga_terjual))
print("Termahal      :", max(harga_terjual))
print("Termurah      :", min(harga_terjual))

rata_rata = sum(harga_terjual) / len(harga_terjual)
print(f"Rata-rata     : {rata_rata:.2f}")
````

````python
# Menambah dan mengurutkan
harga_terjual.append(19.99)              # tambah satu barang di akhir list
print(harga_terjual)
print(sorted(harga_terjual))             # urut kecil -> besar (list asli tidak berubah)
print(sorted(harga_terjual, reverse=True))  # urut besar -> kecil
````

````python
# List bisa berisi teks
status_pesanan = ["Complete", "Shipped", "Complete", "Cancelled", "Processing", "Complete"]
print(status_pesanan.count("Complete"))   # berapa kali "Complete" muncul
print("Returned" in status_pesanan)       # apakah ada "Returned"? -> True/False
````

## 5. Dictionary: data berlabel

Dictionary menyimpan pasangan **key: value** di dalam kurung kurawal `{ }`. Cocok untuk menyimpan satu "baris" data, misalnya satu pesanan:

````python
pesanan = {
    "order_id": 1001,
    "user_id": 57,
    "status": "Complete",
    "num_of_item": 2,
    "total": 89.98,
}
print(pesanan)
````

````python
# Mengambil nilai dengan key
print(pesanan["status"])
print(pesanan["total"])

# Mengubah nilai dan menambah key baru
pesanan["status"] = "Returned"
pesanan["country"] = "Japan"
print(pesanan)
````

````python
print(pesanan.keys())     # semua key
print(pesanan.values())   # semua value
print(pesanan.get("gender", "tidak ada"))   # .get() tidak error kalau key tidak ada
````

### Dictionary untuk ringkasan

Dictionary juga cocok untuk ringkasan seperti "pendapatan per kategori". Ini mirip hasil `GROUP BY` di SQL.

````python
pendapatan_per_kategori = {
    "Outerwear & Coats": 45210.50,
    "Jeans": 38990.25,
    "Sweaters": 30120.00,
    "Swim": 18550.75,
}

total_pendapatan = sum(pendapatan_per_kategori.values())
kategori_terbesar = max(pendapatan_per_kategori, key=pendapatan_per_kategori.get)

print(f"Total pendapatan : {total_pendapatan:,.2f} USD")
print(f"Kategori terbesar: {kategori_terbesar}")

porsi_jeans = pendapatan_per_kategori["Jeans"] / total_pendapatan * 100
print(f"Porsi Jeans      : {porsi_jeans:.1f}%")
````

## 6. List berisi dictionary: tabel kecil

Gabungan keduanya, yaitu list yang setiap isinya dictionary, adalah cara Python menyimpan **tabel**. Setiap dictionary adalah satu baris. Ini cikal bakal DataFrame pandas di sesi 7.

````python
daftar_pesanan = [
    {"order_id": 1001, "status": "Complete",  "country": "Japan",         "total": 89.98},
    {"order_id": 1002, "status": "Shipped",   "country": "United States", "total": 35.50},
    {"order_id": 1003, "status": "Cancelled", "country": "Brasil",        "total": 120.00},
    {"order_id": 1004, "status": "Complete",  "country": "China",         "total": 15.99},
    {"order_id": 1005, "status": "Complete",  "country": "United States", "total": 210.40},
]

print("Jumlah pesanan:", len(daftar_pesanan))
print("Pesanan pertama:", daftar_pesanan[0])
print("Negara pesanan ketiga:", daftar_pesanan[2]["country"])
````

````python
# Menghitung total semua pesanan.
# Untuk sekarang kita tulis satu per satu. Di sesi 6 kita pakai loop
# supaya bisa untuk ribuan pesanan.
total_semua = (daftar_pesanan[0]["total"] + daftar_pesanan[1]["total"]
               + daftar_pesanan[2]["total"] + daftar_pesanan[3]["total"]
               + daftar_pesanan[4]["total"])
print(f"Total semua pesanan: {total_semua:.2f} USD")
````
