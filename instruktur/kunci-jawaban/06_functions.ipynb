## 1. Perbandingan: hasilnya True atau False

| Operator | Arti |
| --- | --- |
| `==` | sama dengan (dua tanda `=`!) |
| `!=` | tidak sama dengan |

````python
total = 85.0
status = "Complete"

print(total > 50)
print(status == "Complete")
print(status == "complete")          # huruf besar-kecil dibedakan
print(total > 50 and status == "Complete")
print(status in ["Cancelled", "Returned"])
````

## 2. if, elif, else: mengambil keputusan

```python
if kondisi_1:
    # dijalankan kalau kondisi_1 benar
elif kondisi_2:
    # dijalankan kalau kondisi_1 salah DAN kondisi_2 benar
else:
    # dijalankan kalau semua kondisi di atas salah
```

**Indentasi (4 spasi atau Tab) itu wajib.** Python memakai indentasi untuk tahu baris mana yang termasuk di dalam `if`. Jangan lupa titik dua `:` di akhir baris `if`, `elif`, dan `else`.

````python
total = 85.0

if total >= 150:
    print("Pesanan besar")
elif total >= 50:
    print("Pesanan sedang")
else:
    print("Pesanan kecil")
````

Coba ganti nilai `total` di atas menjadi `30` lalu `200`, dan jalankan lagi. Persis seperti `CASE WHEN` di SQL, kondisi dicek dari atas ke bawah dan yang pertama benar yang dipakai.

````python
# Contoh lain: apakah pesanan ini dihitung sebagai pendapatan?
status = "Returned"

if status in ["Cancelled", "Returned"]:
    print("Tidak dihitung sebagai pendapatan")
else:
    print("Dihitung sebagai pendapatan")
````

## 3. Loop for: mengulang untuk setiap isi list

```python
for nama_variabel in list:
    # dijalankan sekali untuk setiap isi list
```

````python
harga_terjual = [49.99, 15.00, 120.50, 32.75, 8.99]

for harga in harga_terjual:
    print(f"Barang seharga {harga} USD")
````

### Pola akumulator: menjumlahkan di dalam loop

Mulai dari 0, lalu tambahkan setiap nilai satu per satu. Pola ini dipakai di mana-mana.

````python
total = 0
for harga in harga_terjual:
    total = total + harga        # bisa juga ditulis: total += harga

print(f"Total: {total:.2f} USD")
````

### Loop + if: memilih dan menghitung

Ini data pesanan yang sama seperti sesi 5. Sekarang kita olah dengan loop, sehingga kodenya tetap pendek walaupun pesanannya ribuan.

````python
daftar_pesanan = [
    {"order_id": 1001, "status": "Complete",  "country": "Japan",         "total": 89.98},
    {"order_id": 1002, "status": "Shipped",   "country": "United States", "total": 35.50},
    {"order_id": 1003, "status": "Cancelled", "country": "Brasil",        "total": 120.00},
    {"order_id": 1004, "status": "Complete",  "country": "China",         "total": 15.99},
    {"order_id": 1005, "status": "Complete",  "country": "United States", "total": 210.40},
    {"order_id": 1006, "status": "Returned",  "country": "France",        "total": 64.00},
    {"order_id": 1007, "status": "Processing","country": "China",         "total": 158.25},
]

pendapatan = 0
jumlah_sah = 0
for pesanan in daftar_pesanan:
    if pesanan["status"] not in ["Cancelled", "Returned"]:
        pendapatan += pesanan["total"]
        jumlah_sah += 1

print(f"Pesanan sah : {jumlah_sah}")
print(f"Pendapatan  : {pendapatan:.2f} USD")
print(f"AOV         : {pendapatan / jumlah_sah:.2f} USD")
````

### Menghitung per kelompok dengan dictionary

Ini cara Python melakukan `GROUP BY status` + `COUNT(*)`:

````python
jumlah_per_status = {}

for pesanan in daftar_pesanan:
    status = pesanan["status"]
    if status in jumlah_per_status:
        jumlah_per_status[status] += 1     # status sudah ada: tambah 1
    else:
        jumlah_per_status[status] = 1      # status baru: mulai dari 1

print(jumlah_per_status)
````

### range(): mengulang sejumlah kali

`range(n)` menghasilkan angka 0 sampai n−1. Berguna kalau kita butuh nomor urut.

````python
for i in range(3):
    print("Ulangan ke-", i)

# Nomor urut + isi list sekaligus dengan enumerate()
for nomor, pesanan in enumerate(daftar_pesanan, start=1):
    print(nomor, pesanan["order_id"], pesanan["status"])
````

## 4. Function: membungkus langkah agar bisa dipakai ulang

Kita sudah memakai banyak function bawaan: `print()`, `len()`, `sum()`, `round()`. Sekarang kita membuat sendiri.

```python
def nama_function(parameter1, parameter2):
    """Penjelasan singkat apa yang dikerjakan function ini."""
    hasil = ...            # langkah-langkah
    return hasil           # nilai yang dikembalikan
```

- `def` = mendefinisikan (membuat) function. Kodenya belum dijalankan sampai function **dipanggil**.
- **Parameter** = input. `return` = output.
- Teks dalam `"""..."""` di baris pertama disebut *docstring*: penjelasan function.

````python
def hitung_margin(harga_jual, biaya):
    """Mengembalikan margin (harga jual - biaya) dalam USD."""
    return harga_jual - biaya

print(hitung_margin(59.90, 27.30))
print(hitung_margin(120.00, 55.00))
````

````python
# Parameter dengan nilai default: dipakai kalau tidak diisi
def harga_setelah_diskon(harga, persen_diskon=10):
    """Mengembalikan harga setelah diskon, dibulatkan 2 desimal."""
    return round(harga * (1 - persen_diskon / 100), 2)

print(harga_setelah_diskon(80))        # diskon default 10%
print(harga_setelah_diskon(80, 25))    # diskon 25%
````

> **`print` vs `return`.** `print` hanya menampilkan ke layar. `return` mengembalikan nilai yang bisa disimpan di variabel dan dipakai lagi. Function untuk analisis hampir selalu memakai `return`.

## 5. Studi kasus: label ukuran pesanan

Tim marketing ingin membagi pesanan menjadi tiga kelompok untuk program ongkir gratis:

| Label | Total pesanan |
| --- | --- |
| `kecil` | di bawah 50 USD |
| `sedang` | 50 sampai di bawah 150 USD |
| `besar` | 150 USD ke atas |

````python
def label_ukuran_pesanan(total):
    """Memberi label 'kecil', 'sedang', atau 'besar' berdasarkan total pesanan (USD)."""
    if total >= 150:
        return "besar"
    elif total >= 50:
        return "sedang"
    else:
        return "kecil"

# Uji dengan beberapa nilai, termasuk nilai di batas
print(label_ukuran_pesanan(20))
print(label_ukuran_pesanan(50))     # tepat di batas: harus "sedang"
print(label_ukuran_pesanan(149.99))
print(label_ukuran_pesanan(150))    # tepat di batas: harus "besar"
````

Sekarang terapkan ke semua pesanan, dan hitung jumlah pesanan per label:

````python
jumlah_per_label = {"kecil": 0, "sedang": 0, "besar": 0}

for pesanan in daftar_pesanan:
    label = label_ukuran_pesanan(pesanan["total"])
    pesanan["ukuran"] = label            # simpan label ke dalam pesanan
    jumlah_per_label[label] += 1

print(jumlah_per_label)
for pesanan in daftar_pesanan:
    print(pesanan["order_id"], pesanan["total"], pesanan["ukuran"])
````

### Function yang mengembalikan beberapa nilai

Function bisa mengembalikan dictionary, sehingga satu panggilan menghasilkan ringkasan lengkap:

````python
def ringkas_pesanan(daftar):
    """Mengembalikan jumlah pesanan sah, pendapatan, dan AOV dari list pesanan."""
    pendapatan = 0
    jumlah = 0
    for p in daftar:
        if p["status"] not in ["Cancelled", "Returned"]:
            pendapatan += p["total"]
            jumlah += 1
    aov = pendapatan / jumlah if jumlah > 0 else 0   # hindari pembagian dengan nol
    return {"jumlah_pesanan": jumlah, "pendapatan": round(pendapatan, 2), "aov": round(aov, 2)}

ringkas_pesanan(daftar_pesanan)
````

**PR1.** Tulis function `kelompok_umur(umur)` yang mengembalikan label yang **sama persis** dengan `CASE WHEN` di sesi 4: `'1. Di bawah 18'`, `'2. 18-24'`, `'3. 25-34'`, `'4. 35-44'`, `'5. 45-54'`, `'6. 55+'`. Uji dengan umur 15, 18, 24, 25, 54, dan 70.

````python
# PR1
def kelompok_umur(umur):
    """Label kelompok umur, sama dengan CASE WHEN di sesi 4."""
    if umur < 18:
        return "1. Di bawah 18"
    elif umur < 25:
        return "2. 18-24"
    elif umur < 35:
        return "3. 25-34"
    elif umur < 45:
        return "4. 35-44"
    elif umur < 55:
        return "5. 45-54"
    else:
        return "6. 55+"

for u in [15, 18, 24, 25, 54, 70]:
    print(u, kelompok_umur(u))
````

**PR2.** Tulis function `persen_margin(harga_jual, biaya)` yang mengembalikan persentase margin dibulatkan 1 desimal. Jika `harga_jual` bernilai 0, kembalikan `0` (jangan sampai error). Uji dengan `(60, 27)` dan `(0, 10)`.

````python
# PR2
def persen_margin(harga_jual, biaya):
    """Persentase margin, 1 desimal. Mengembalikan 0 jika harga_jual 0."""
    if harga_jual == 0:
        return 0
    return round((harga_jual - biaya) / harga_jual * 100, 1)

print(persen_margin(60, 27), persen_margin(0, 10))
````

**PR3.** Dari `daftar_pesanan`, gunakan loop untuk menghitung **pendapatan per negara** (hanya pesanan sah) dan simpan di dictionary. Negara mana yang pendapatannya terbesar?

````python
# PR3
# Cara dengan if/else seperti demo; .get(key, 0) juga benar
pendapatan_per_negara = {}
for pesanan in daftar_pesanan:
    if pesanan["status"] not in ["Cancelled", "Returned"]:
        negara = pesanan["country"]
        if negara in pendapatan_per_negara:
            pendapatan_per_negara[negara] += pesanan["total"]
        else:
            pendapatan_per_negara[negara] = pesanan["total"]
print(pendapatan_per_negara)
print("Terbesar:", max(pendapatan_per_negara, key=pendapatan_per_negara.get))
````
